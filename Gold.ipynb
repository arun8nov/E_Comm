```python
%sql 

CREATE OR REPLACE TABLE e_comm.gold_e_comm.gld_products AS
WITH brands_categories AS(
  SELECT 
    b.brand_name,
    b.brand_code,
    c.category_name,
    c.Category_code
  FROM brands b
  INNER JOIN categories c
  ON b.category_code = c.category_code
)
SELECT
  p.product_id,
  p.sku,
  p.category_code,
  COALESCE( bc.category_name, 'Not Available') AS category_name,
  p.brand_code,
  COALESCE( bc.brand_name, 'Not Available') AS brand_name,
  p.color,
  p.size,
  p.material,
  p.weight_grams,
  p.length_cm,
  p.width_cm,
  p.height_cm,
  p.rating_count,
  p.file_name,
  p.ingested_at
FROM products p
LEFT JOIN brands_categories bc
  ON p.brand_code = bc.brand_code
```

```python
%sql
WITH regions AS(
SELECT 
  country,
  state,
  COUNT(*)
FROM customers
GROUP BY country, state
ORDER BY country
)
SELECT
  country,
  state,
  CASE
    WHEN country = 'Australia' AND state = 'NSW' THEN 'EAST'
    WHEN country = 'Australia' AND state = 'QLD' THEN 'NorthEast'
    WHEN country = 'Australia' AND state = 'VIC' THEN 'SouthEast'
    WHEN country = 'Australia' AND state = 'WA' THEN 'West'

    WHEN country = 'Canada' AND state = 'BC' THEN 'West'
    WHEN country = 'Canada' AND state = 'QC' THEN 'East'
    WHEN country = 'Canada' AND state = 'ON' THEN 'East'
    WHEN country = 'Canada' AND state = 'AB' THEN 'West'

    WHEN country = 'India' AND state = 'WB' THEN 'North'
    WHEN country = 'India' AND state = 'GJ' THEN 'West'
    WHEN country = 'India' AND state = 'MH' THEN 'West'
    WHEN country = 'India' AND state = 'KL' THEN 'South'
    WHEN country = 'India' AND state = 'TS' THEN 'South'
    WHEN country = 'India' AND state = 'KA' THEN 'South'
    WHEN country = 'India' AND state = 'TN' THEN 'South'
    WHEN country = 'India' AND state = 'RJ' THEN 'West'
    WHEN country = 'India' AND state = 'UP' THEN 'North'
    WHEN country = 'India' AND state = 'DL' THEN 'North'

    WHEN country = 'Singapore' AND state = 'SG' THEN 'Singapore'

    WHEN country = 'United Arab Emirates' AND state = 'DU' THEN 'Dubai'
    WHEN country = 'United Arab Emirates' AND state = 'AUH' THEN 'Abu Dhabi'
    WHEN country = 'United Arab Emirates' AND state = 'SHJ' THEN 'Sharjah'
    
    WHEN country = 'United Kingdom' AND state = 'NIR' THEN 'Northern Ireland'
    WHEN country = 'United Kingdom' AND state = 'ENG' THEN 'England'
    WHEN country = 'United Kingdom' AND state = 'SCT' THEN 'Scotland'
    WHEN country = 'United Kingdom' AND state = 'WLS' THEN 'Wales'

    WHEN country = 'United States' AND state = 'CA' THEN 'West'
    WHEN country = 'United States' AND state = 'NY' THEN 'NorthEast'
    WHEN country = 'United States' AND state = 'MA' THEN 'NorthEast'
    WHEN country = 'United States' AND state = 'FL' THEN 'South'
    WHEN country = 'United States' AND state = 'WA' THEN 'Northwest'
    WHEN country = 'United States' AND state = 'NJ' THEN 'NorthEast'
    WHEN country = 'United States' AND state = 'TX' THEN 'South'
    WHEN country = 'United States' AND state = 'IL' THEN 'Midwest regio'
    
  END AS region
FROM regions

```

```python
%sql

CREATE OR REPLACE TABLE e_comm.gold_e_comm.gld_customers AS

WITH Regions AS(
WITH regions AS(
SELECT 
  country,
  state,
  COUNT(*)
FROM customers
GROUP BY country, state
ORDER BY country
)
SELECT
  country,
  state,
  CASE
    WHEN country = 'Australia' AND state = 'NSW' THEN 'EAST'
    WHEN country = 'Australia' AND state = 'QLD' THEN 'NorthEast'
    WHEN country = 'Australia' AND state = 'VIC' THEN 'SouthEast'
    WHEN country = 'Australia' AND state = 'WA' THEN 'West'

    WHEN country = 'Canada' AND state = 'BC' THEN 'West'
    WHEN country = 'Canada' AND state = 'QC' THEN 'East'
    WHEN country = 'Canada' AND state = 'ON' THEN 'East'
    WHEN country = 'Canada' AND state = 'AB' THEN 'West'

    WHEN country = 'India' AND state = 'WB' THEN 'North'
    WHEN country = 'India' AND state = 'GJ' THEN 'West'
    WHEN country = 'India' AND state = 'MH' THEN 'West'
    WHEN country = 'India' AND state = 'KL' THEN 'South'
    WHEN country = 'India' AND state = 'TS' THEN 'South'
    WHEN country = 'India' AND state = 'KA' THEN 'South'
    WHEN country = 'India' AND state = 'TN' THEN 'South'
    WHEN country = 'India' AND state = 'RJ' THEN 'West'
    WHEN country = 'India' AND state = 'UP' THEN 'North'
    WHEN country = 'India' AND state = 'DL' THEN 'North'

    WHEN country = 'Singapore' AND state = 'SG' THEN 'Singapore'

    WHEN country = 'United Arab Emirates' AND state = 'DU' THEN 'Dubai'
    WHEN country = 'United Arab Emirates' AND state = 'AUH' THEN 'Abu Dhabi'
    WHEN country = 'United Arab Emirates' AND state = 'SHJ' THEN 'Sharjah'
    
    WHEN country = 'United Kingdom' AND state = 'NIR' THEN 'Northern Ireland'
    WHEN country = 'United Kingdom' AND state = 'ENG' THEN 'England'
    WHEN country = 'United Kingdom' AND state = 'SCT' THEN 'Scotland'
    WHEN country = 'United Kingdom' AND state = 'WLS' THEN 'Wales'

    WHEN country = 'United States' AND state = 'CA' THEN 'West'
    WHEN country = 'United States' AND state = 'NY' THEN 'NorthEast'
    WHEN country = 'United States' AND state = 'MA' THEN 'NorthEast'
    WHEN country = 'United States' AND state = 'FL' THEN 'South'
    WHEN country = 'United States' AND state = 'WA' THEN 'Northwest'
    WHEN country = 'United States' AND state = 'NJ' THEN 'NorthEast'
    WHEN country = 'United States' AND state = 'TX' THEN 'South'
    WHEN country = 'United States' AND state = 'IL' THEN 'Midwest regio'
    
  END AS region
FROM regions)

SELECT 
  c.customer_id,
  c.phone,
  c.country_code,
  c.country,
  c.state,
  COALESCE(R.region, 'Not Available') AS region,
  c.file_name,
  c.ingested_at
  FROM customers c
  LEFT JOIN Regions R
    ON c.country = R.country
    AND c.state = R.state


```
